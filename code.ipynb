```python
import pandas as pd
import sqlite3


orders = pd.read_csv(r"C:\Users\guthp\Downloads\orders.csv")
users = pd.read_json(r"C:\Users\guthp\Downloads\users.json")
conn = sqlite3.connect("restaurants.db")  

with open(r"C:\Users\guthp\Downloads\restaurants.sql", "r") as f:
    sql_script = f.read()

conn.executescript(sql_script)

restaurants = pd.read_sql("SELECT * FROM restaurants", conn)

merged = orders.merge(users, on="user_id", how="left")
final_df = merged.merge(restaurants, on="restaurant_id", how="left")

final_df.to_csv("final_food_delivery_dataset.csv", index=False)
print("Final dataset created successfully ")
```
